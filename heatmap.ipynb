```python
import matplotlib.pyplot as plt
import numpy as np
import os
import pandas as pd

# Data for plotting
for i in range(1,26):
    path = f"~/ray_results/param_search/config{i}/"
    folder_name = os.listdir(os.path.expanduser(path))[0]


    
    file_path = os.path.join(os.path.expanduser(path), folder_name, "progress.csv")
    
    with open(file_path, 'r') as f:
        df = pd.read_csv(f)

    # Extract entropy, learning rate, and reward from the data

    entropy = df["info/learner/mediator_policy/learner_stats/entropy_coeff"].iloc[-1]
    learning_rate = df["info/learner/mediator_policy/learner_stats/cur_lr"].iloc[-1]
    reward = df["episode_reward_max"].iloc[-1]

    # Store the values in a list
    if i == 0:
        entropies = []
        learning_rates = []
        rewards = []

    entropies.append(entropy)
    learning_rates.append(learning_rate)
    rewards.append(reward)

# Create a 2D grid for entropy and learning rate
entropy_unique = np.unique(entropies)
learning_rate_unique = np.unique(learning_rates)
reward_matrix = np.zeros((len(entropy_unique), len(learning_rate_unique)))

# Fill the reward matrix
for e, lr, r in zip(entropies, learning_rates, rewards):
    e_idx = np.where(entropy_unique == e)[0][0]
    lr_idx = np.where(learning_rate_unique == lr)[0][0]
    reward_matrix[e_idx, lr_idx] = r

# Plot the heatmap
plt.figure(figsize=(10, 8))
plt.imshow(reward_matrix, cmap='viridis', origin='lower', aspect='auto')
plt.colorbar(label='Reward')
plt.xticks(ticks=np.arange(len(learning_rate_unique)), labels=np.round(learning_rate_unique, 5), rotation=45)
plt.yticks(ticks=np.arange(len(entropy_unique)), labels=np.round(entropy_unique, 5))
plt.xlabel('Learning Rate')
plt.ylabel('Entropy Coefficient')
plt.title('Heatmap of Reward vs Entropy and Learning Rate')
plt.show()
```
